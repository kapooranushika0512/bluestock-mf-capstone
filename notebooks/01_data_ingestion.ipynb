```python
import pandas as pd

funds = pd.read_csv("../data/raw/01_fund_master.csv")

print("Fund Houses")
print(funds["fund_house"].unique())

print("\nCategories")
print(funds["category"].unique())

print("\nSub Categories")
print(funds["sub_category"].unique())

print("\nRisk Categories")
print(funds["risk_category"].unique())
```

```python
print("Total Funds:", len(funds))

print("\nFund House Counts:")
print(funds["fund_house"].value_counts())
```

```python
nav_history = pd.read_csv("../data/raw/02_nav_history.csv")

missing_codes = set(funds["amfi_code"]) - set(nav_history["amfi_code"])

print("Missing Codes:", missing_codes)
print("Total Missing:", len(missing_codes))
```
